## 1.2 Accessing List Elements

Lists use **zero-based indexing**. You can also use **negative indexing** to access from the end.

```
 Index:    0      1       2       3
 List:  ["apple", "banana", "cherry", "mango"]
 Neg:     -4      -3       -2       -1
```

````python
# ── Accessing List Elements ──────────────────────────────────────

fruits = ["apple", "banana", "cherry", "mango", "orange"]

# Positive indexing
print("First element  (index 0):", fruits[0])
print("Second element (index 1):", fruits[1])
print("Third element  (index 2):", fruits[2])

print("element at index 3:", fruits[3])
````

````python
# Negative indexing
print("\nLast element    (index -1):", fruits[-1])
print("Second last     (index -2):", fruits[-2])

````

````python
# Slicing: list[start:stop:step]
print("\n--- Slicing ---")
print(fruits)
print("fruits[1:4]   :", fruits[1:4])   # index 1 to 3
print("fruits[:3]    :", fruits[:3])    # first 3
print("fruits[2:]    :", fruits[2:])    # from index 2 to end
print("fruits[::2]   :", fruits[::2])   # every 2nd element
print("fruits[:4:2]  :", fruits[:4:2])  # from index 0 to 3, every 2nd element
print("fruits[::-1]  :", fruits[::-1])  # reversed

````

## What is a Tuple?
A **Tuple** is an **ordered**, **immutable** (unchangeable), and **indexed** collection. Once created, its elements **cannot be modified**. Tuples are faster than lists and are used to store data that should not change.

### Key Characteristics:
| Property | Value |
|---|---|
| Ordered | ✅ Yes |
| Mutable | ❌ No (Immutable) |
| Allows Duplicates | ✅ Yes |
| Indexed | ✅ Yes |
| Syntax | `( )` |

````python
# ── 🎯 Practical Program: GPS Coordinate Tracker ─────────────────

print("=" * 45)
print("       GPS COORDINATE TRACKER")
print("=" * 45)

# Storing locations as tuples (immutable — coordinates don't change!)
locations = {
    "Home"      : (17.3850, 78.4867),
    "College"   : (17.4400, 78.4983),
    "Hospital"  : (17.3616, 78.4747),
    "Airport"   : (17.2403, 78.4294),
}

import math

def distance(loc1, loc2):
    """Approximate distance in km between two lat/lon points."""
    lat1, lon1 = loc1
    lat2, lon2 = loc2
    return math.sqrt((lat2 - lat1)**2 + (lon2 - lon1)**2) * 111

print("\n📍 Saved Locations:")
for name, coords in locations.items():
    lat, lon = coords   # tuple unpacking
    print(f"  {name:<12}: Lat={lat}, Lon={lon}")

home = locations["Home"]
print("\n📏 Distances from Home:")
for name, coords in locations.items():
    if name != "Home":
        d = distance(home, coords)
        print(f"  Home → {name:<12}: {d:.2f} km")
````

## What is a Set?

### Key Characteristics:

## 3.2 All Set Methods — Explained

### Modification Methods:
| Method | Description |
|---|---|
| `add(x)` | Adds element `x` |
| `update(iterable)` | Adds multiple elements |
| `remove(x)` | Removes `x`; raises error if not found |
| `discard(x)` | Removes `x`; no error if not found |
| `pop()` | Removes and returns an arbitrary element |
| `clear()` | Removes all elements |

### Set Operations (Mathematical):
| Method | Operator | Description |
|---|---|---|
| `union(s)` | `\|` | All elements from both sets |
| `intersection(s)` | `&` | Elements common to both |
| `difference(s)` | `-` | Elements in first but not second |
| `symmetric_difference(s)` | `^` | Elements in either but not both |
| `issubset(s)` | `<=` | Is this set a subset? |
| `issuperset(s)` | `>=` | Is this set a superset? |
| `isdisjoint(s)` | — | Do sets share no elements? |

# 5. DIFFERENCES AMONG DATA STRUCTURES 📊

## Comprehensive Comparison Table

| Feature | List `[ ]` | Tuple `( )` | Set `{ }` | Dictionary `{k:v}` |
|---|---|---|---|---|
| **Ordered** | ✅ Yes | ✅ Yes | ❌ No | ✅ Yes (3.7+) |
| **Mutable** | ✅ Yes | ❌ No | ✅ Yes | ✅ Yes |
| **Duplicates** | ✅ Yes | ✅ Yes | ❌ No | Keys: ❌ / Values: ✅ |
| **Indexed** | ✅ Yes | ✅ Yes | ❌ No | By key |
| **Syntax** | `[1,2,3]` | `(1,2,3)` | `{1,2,3}` | `{'a':1}` |
| **Access** | By index | By index | Membership only | By key |
| **Speed** | Moderate | Fastest | Very Fast | Fast |
| **Use Case** | General purpose | Fixed data | Unique items | Key-value pairs |

````python
# ── Side-by-Side Comparison ───────────────────────────────────────

# Same data in all four structures
my_list  = [1, 2, 3, 2, 1]          # allows duplicates
my_tuple = (1, 2, 3, 2, 1)          # immutable
my_set   = {1, 2, 3, 2, 1}          # removes duplicates → {1,2,3}
my_dict  = {"a": 1, "b": 2, "c": 3} # key-value pairs

print("List  :", my_list)
print("Tuple :", my_tuple)
print("Set   :", my_set)    # notice: no duplicates, unordered
print("Dict  :", my_dict)

# Mutability test
my_list[0] = 99
print("\nList after edit [0]=99:", my_list)

try:
    my_tuple[0] = 99  # This will FAIL
except TypeError as e:
    print("Tuple modification error:", e)

# Memory comparison (tuples are more efficient)
import sys
lst = list(range(100))
tpl = tuple(range(100))
print(f"\nMemory — List of 100: {sys.getsizeof(lst)} bytes")
print(f"Memory — Tuple of 100: {sys.getsizeof(tpl)} bytes")
print("→ Tuples use less memory!")

# Speed test
import timeit
list_time  = timeit.timeit("[1,2,3,4,5]", number=1_000_000)
tuple_time = timeit.timeit("(1,2,3,4,5)", number=1_000_000)
print(f"\nCreation speed (1M times):")
print(f"  List  = {list_time:.4f}s")
print(f"  Tuple = {tuple_time:.4f}s")
print("→ Tuples are faster to create!")
````

# 7. BRANCHING AND CONTROL LOOPS WITH DATA STRUCTURES 🔄

## 7.1 if / elif / else — Branching

Branching allows your program to make decisions based on conditions.

```
if condition:
    # runs when condition is True
elif another_condition:
    # runs when second condition is True
else:
    # runs when no condition is True
```

````python
# ── Branching with Data Structures ──────────────────────────────

# 1. if-elif-else with a List
print("--- Grade Classifier (List + if-elif-else) ---")
marks = [92, 45, 75, 55, 83, 38, 67]

for m in marks:
    if m >= 90:
        grade = "A+ (Outstanding)"
    elif m >= 75:
        grade = "A  (Excellent)"
    elif m >= 60:
        grade = "B  (Good)"
    elif m >= 50:
        grade = "C  (Average)"
    else:
        grade = "F  (Fail)"
    print(f"  Marks: {m:>3} → {grade}")

# 2. Branching with Dictionary
print("\n--- Menu System (Dictionary + if-elif) ---")
menu = {1: "Start Game", 2: "Load Game", 3: "Settings", 4: "Exit"}

choice = 2
if choice in menu:
    print(f"  You selected: {menu[choice]}")
else:
    print("  Invalid choice!")

# 3. Branching with Set (membership)
print("\n--- Access Control (Set + if) ---")
admins = {"alice", "bob", "carol"}
users  = ["alice", "dave", "bob", "eve"]

for user in users:
    if user in admins:
        print(f"  ✅ {user:<8} — Admin Access Granted")
    else:
        print(f"  ❌ {user:<8} — Regular User")
````

## 7.2 for / while Loops

| Loop | Use When |
|---|---|
| `for` | You know how many iterations (iterating over a collection) |
| `while` | You loop until a condition becomes False |

### Loop Control Statements:
| Statement | Effect |
|---|---|
| `break` | Exits the loop immediately |
| `continue` | Skips to the next iteration |
| `pass` | Does nothing (placeholder) |

````python
# ── Loops with Data Structures ───────────────────────────────────

# FOR loop over List
print("--- for loop over List ---")
fruits = ["apple", "banana", "cherry", "date"]
for i, fruit in enumerate(fruits, start=1):
    print(f"  {i}. {fruit.capitalize()}")

# FOR loop over Dictionary
print("\n--- for loop over Dictionary ---")
scores = {"Alice": 92, "Bob": 75, "Charlie": 88}
for name, score in scores.items():
    status = "PASS" if score >= 80 else "FAIL"
    print(f"  {name:<10}: {score}  [{status}]")

# FOR loop over Set
print("\n--- for loop over Set ---")
unique_tags = {"python", "data-science", "ml", "ai"}
for tag in sorted(unique_tags):
    print(f"  #{tag}")

# WHILE loop with List
print("\n--- while loop draining a stack ---")
stack = [10, 20, 30, 40, 50]
while stack:
    item = stack.pop()
    print(f"  Processed: {item}  | Remaining: {stack}")

# break and continue with List
print("\n--- break: stop at first multiple of 7 ---")
numbers = [3, 14, 9, 21, 8, 35, 6]
for n in numbers:
    if n % 7 == 0:
        print(f"  Found multiple of 7: {n} → stopping!")
        break
    print(f"  Checked: {n}")

print("\n--- continue: skip even numbers ---")
for n in range(1, 11):
    if n % 2 == 0:
        continue
    print(f"  Odd: {n}")
````

## 8.3 Element-wise & Matrix Arithmetic

````python
# ── Element-wise Arithmetic ──────────────────────────────────────

A = np.array([[1, 2], [3, 4]])
B = np.array([[5, 6], [7, 8]])

print("A:\n", A)
print("B:\n", B)

print("\nA + B  (element-wise add):\n",  A + B)
print("A - B  (element-wise sub):\n",  A - B)
print("A * B  (element-wise mul):\n",  A * B)
print("A / B  (element-wise div):\n",  A / B)
print("A ** 2 (element-wise power):\n", A ** 2)

# Scalar operations
print("\nA + 10 (scalar add):\n", A + 10)
print("A * 3  (scalar mul):\n", A * 3)

# Useful aggregate functions
C = np.array([[3, 1, 4], [1, 5, 9], [2, 6, 5]])
print("\nMatrix C:\n", C)
print("sum()      :", C.sum())
print("sum(axis=0):", C.sum(axis=0))   # column sums
print("sum(axis=1):", C.sum(axis=1))   # row sums
print("mean()     :", C.mean())
print("std()      :", round(C.std(), 4))
print("min(), max():", C.min(), C.max())
print("argmax()   :", C.argmax())       # flat index of max
````

## 8.4 Linear Algebra Operations

````python
# ── Linear Algebra with NumPy ────────────────────────────────────

A = np.array([[2, 1], [5, 3]])
B = np.array([[1, 0], [2, 1]])

print("A:\n", A)
print("B:\n", B)

# Matrix multiplication (dot product)
print("\nMatrix Multiplication A @ B:\n", A @ B)
print("(same as np.dot(A, B)):\n", np.dot(A, B))

# Transpose
print("\nTranspose of A (A.T):\n", A.T)

# Determinant
det_A = np.linalg.det(A)
print("\nDeterminant of A:", det_A)

# Inverse
inv_A = np.linalg.inv(A)
print("Inverse of A:\n", inv_A)
print("Verify A @ inv(A) ≈ I:\n", np.round(A @ inv_A, 10))

# Eigenvalues and Eigenvectors
eigenvalues, eigenvectors = np.linalg.eig(A)
print("\nEigenvalues :", eigenvalues)
print("Eigenvectors:\n", eigenvectors)

# Solving linear equations: Ax = b
# 2x + y = 5
# 5x + 3y = 13
b = np.array([5, 13])
x = np.linalg.solve(A, b)
print("\nSolving Ax=b where b=[5,13]:")
print("x =", x)   # should be x=2, y=1
print("Verification A@x:", A @ x)
````

## 8.5 Reshaping & Stacking

````python
# ── Reshape, Stack, Split ────────────────────────────────────────

a = np.arange(1, 13)   # [1..12]
print("Original:", a)

# reshape()
m = a.reshape(3, 4)    # 3 rows, 4 columns
print("\nreshaped to (3,4):\n", m)

m2 = a.reshape(2, 2, 3)  # 3D tensor
print("\nreshaped to (2,2,3):\n", m2)

# flatten() — back to 1D
flat = m.flatten()
print("\nflattened:", flat)

# Stacking
r1 = np.array([1, 2, 3])
r2 = np.array([4, 5, 6])

vstacked = np.vstack([r1, r2])   # vertical stack (rows)
hstacked = np.hstack([r1, r2])   # horizontal stack

print("\nvstack:\n", vstacked)
print("hstack:", hstacked)

# Splitting
splits = np.split(np.arange(12), 3)   # split into 3 equal parts
print("\nSplit into 3 parts:", splits)
````

````python
# ── 🎯 Practical Program: Student Marks Matrix Analysis ───────────

print("=" * 60)
print("       STUDENT EXAM MARKS — MATRIX ANALYSIS")
print("=" * 60)

import numpy as np

students = ["Alice", "Bob", "Charlie", "Diana", "Eve"]
subjects = ["Math", "Physics", "Chemistry", "English", "CS"]

# Marks matrix: rows = students, cols = subjects
marks = np.array([
    [88, 92, 79, 85, 95],   # Alice
    [72, 68, 75, 80, 70],   # Bob
    [95, 98, 92, 88, 99],   # Charlie
    [60, 55, 62, 70, 58],   # Diana
    [84, 79, 88, 91, 87],   # Eve
])

print(f"\n{'Student':<12}", end="")
for s in subjects:
    print(f"{s:>10}", end="")
print(f"{'Total':>10} {'Avg':>8} {'Grade':>8}")
print("-" * 75)

totals = marks.sum(axis=1)
avgs   = marks.mean(axis=1)

for i, student in enumerate(students):
    row = marks[i]
    avg = avgs[i]
    grade = "A+" if avg>=90 else "A" if avg>=80 else "B" if avg>=70 else "C" if avg>=60 else "F"
    print(f"{student:<12}", end="")
    for m in row:
        print(f"{m:>10}", end="")
    print(f"{totals[i]:>10} {avg:>8.1f} {grade:>8}")

print("-" * 75)

# Subject-wise statistics
print("\n📊 Subject-wise Statistics:")
col_avg = marks.mean(axis=0)
col_max = marks.max(axis=0)
col_min = marks.min(axis=0)

print(f"  {'Subject':<12} {'Avg':>8} {'Max':>8} {'Min':>8}")
for j, sub in enumerate(subjects):
    print(f"  {sub:<12} {col_avg[j]:>8.1f} {col_max[j]:>8} {col_min[j]:>8}")

# Topper
topper_idx = avgs.argmax()
print(f"\n🏆 Overall Topper: {students[topper_idx]} ({avgs[topper_idx]:.1f} avg)")

# Students below 70 average
below = [students[i] for i in range(len(students)) if avgs[i] < 70]
print(f"⚠️  Below 70 avg  : {below if below else 'None'}")
````

````python
# ── 💡 Bonus: Exercise Solutions Starter Code ────────────────────

# Exercise 5: Character frequency counter
text = "python programming"
char_freq = {}
for ch in text:
    if ch != " ":
        char_freq[ch] = char_freq.get(ch, 0) + 1

print("Character Frequencies:")
for ch, freq in sorted(char_freq.items(), key=lambda x: -x[1]):
    print(f"  '{ch}': {'▮' * freq} ({freq})")

print()

# Exercise 6: Normalize a NumPy matrix
import numpy as np
M = np.random.randint(1, 100, size=(5, 5)).astype(float)
normalized = (M - M.min()) / (M.max() - M.min())
rank = np.linalg.matrix_rank(M)

print("Random 5x5 Matrix:\n", M)
print("\nNormalized (0 to 1):\n", np.round(normalized, 3))
print("\nMatrix Rank:", rank)
````
