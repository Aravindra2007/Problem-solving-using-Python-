## 📌 Topic 1: Pre-defined Functions (Built-in Functions)

### What are Pre-defined Functions?
Python comes with a large collection of **built-in functions** that are always available — no import needed.  
These are also called **pre-defined functions** because they are already written and stored inside Python's interpreter.

### Why use them?
- Saves time — no need to write common logic from scratch
- Highly optimized and tested
- Available globally in any Python program

### Commonly Used Pre-defined Functions:

| Function | Purpose | Example |
|----------|---------|---------|
| `print()` | Display output | `print("Hello")` |
| `input()` | Read user input | `input("Enter: ")` |
| `len()` | Length of a sequence | `len([1,2,3])` → 3 |
| `type()` | Data type of a value | `type(3.14)` → float |
| `int()`, `float()`, `str()` | Type conversion | `int("10")` → 10 |
| `max()`, `min()` | Max/min of sequence | `max([3,1,5])` → 5 |
| `sum()` | Sum of iterable | `sum([1,2,3])` → 6 |
| `abs()` | Absolute value | `abs(-7)` → 7 |
| `round()` | Round a number | `round(3.567, 2)` → 3.57 |
| `sorted()` | Returns sorted list | `sorted([3,1,2])` → [1,2,3] |
| `range()` | Generates a range | `range(1, 6)` |
| `enumerate()` | Index + value pairs | `enumerate(['a','b'])` |
| `zip()` | Combine iterables | `zip([1,2],[3,4])` |

## 📌 Topic 2: User-Defined Functions

### What is a User-Defined Function?
A **user-defined function** is a block of reusable code that you write yourself to perform a specific task.

### Why Define Functions?
- **Reusability** — Write once, use many times
- **Modularity** — Break large problems into smaller pieces
- **Readability** — Code is easier to understand
- **Debugging** — Errors are easier to isolate

### Syntax:
```python
def function_name(parameters):
    """Docstring: describes what the function does"""
    # body of function
    return value   # optional
```

### Key Terms:
- `def` — keyword to define a function
- `function_name` — identifier following naming rules
- `parameters` — inputs the function expects (optional)
- `docstring` — description string (best practice)
- `return` — sends a result back (optional)

### Types of User-Defined Functions:
| Type | Description |
|------|-------------|
| No parameter, No return | Just performs an action |
| With parameters, No return | Takes input, no output |
| No parameter, With return | Returns a computed value |
| With parameters, With return | Full input-output function |

---


````python
# ═══════════════════════════════════════════════════
# PROGRAM 2.1 — Four Types of User-Defined Functions
# ═══════════════════════════════════════════════════

# Type 1: No parameter, No return
def greet():
    print("\n👋 Welcome to Python Functions!")

# Type 2: With parameter, No return
def display_student(name, roll):
    print(f"\n📋 Student: {name}  |  Roll No: {roll}")

# Type 3: No parameter, With return
def get_pi():
    return 3.14159265358979

# Type 4: With parameter, With return
def calculate_circle_area(radius):
    """Returns the area of a circle given its radius."""
    pi = get_pi()        # calling another function inside!
    return pi * radius * radius

# ── Calling all four types ──
greet()
display_student("Alice", "CS101")
print(f"\n🔵 Value of Pi     : {get_pi()}")
area = calculate_circle_area(7)
print(f"🔵 Area of Circle  : {round(area, 4)} sq. units")
````

````python
# ═══════════════════════════════════════════════════
# PROGRAM 2.2 — Real-World Function: BMI Calculator
# ═══════════════════════════════════════════════════

def calculate_bmi(weight_kg, height_m):
    """
    Calculates Body Mass Index (BMI).
    Formula: BMI = weight / (height^2)
    """
    bmi = weight_kg / (height_m ** 2)
    return round(bmi, 2)

def classify_bmi(bmi):
    """Returns health classification based on BMI value."""
    if bmi < 18.5:
        return "Underweight 🔵"
    elif bmi < 25:
        return "Normal Weight 🟢"
    elif bmi < 30:
        return "Overweight 🟡"
    else:
        return "Obese 🔴"

def bmi_report(name, weight, height):
    """Generates a full BMI report for a person."""
    bmi   = calculate_bmi(weight, height)
    label = classify_bmi(bmi)
    print(f"\n{'='*40}")
    print(f"  BMI REPORT — {name}")
    print(f"{'='*40}")
    print(f"  Weight    : {weight} kg")
    print(f"  Height    : {height} m")
    print(f"  BMI Score : {bmi}")
    print(f"  Status    : {label}")
    print(f"{'='*40}")

# Test the system
bmi_report("Alice",  55, 1.65)
bmi_report("Bob",    90, 1.75)
bmi_report("Charlie",40, 1.70)
````

## 📌 Topic 3: Formal and Actual Parameters

### Definitions:
- **Formal Parameters** — The variable names listed inside the `def` statement. They act as **placeholders**.
- **Actual Parameters (Arguments)** — The real values passed when **calling** the function.

```python
def add(a, b):       # a, b are FORMAL parameters
    return a + b

add(10, 20)          # 10, 20 are ACTUAL parameters (arguments)
```

### Types of Arguments in Python:

| Type | Description | Example |
|------|-------------|---------|
| **Positional** | Matched by position | `func(10, 20)` |
| **Keyword** | Matched by name | `func(b=20, a=10)` |
| **Default** | Pre-set value if not passed | `def func(x, n=2)` |
| **Variable-length (`*args`)** | Accepts any number of positional args | `def func(*args)` |
| **Keyword Variable-length (`**kwargs`)** | Accepts any number of keyword args | `def func(**kwargs)` |

---


````python
# ═══════════════════════════════════════════════════
# PROGRAM 3.1 — Positional, Keyword and Default Arguments
# ═══════════════════════════════════════════════════

def student_fee(name, course, fee=15000, scholarship=0):
    """Calculates net fee after scholarship deduction."""
    net = fee - scholarship
    print(f"  {name:<12} | {course:<10} | ₹{fee:<8} | ₹{scholarship:<8} | ₹{net}")

print("\nName         | Course     | Fee       | Scholar   | Net Fee")
print("-" * 65)

# Positional arguments
student_fee("Alice", "BCA")

# Keyword arguments (order doesn't matter)
student_fee(course="B.Tech", name="Bob", fee=25000)

# Overriding default
student_fee("Charlie", "MCA", 20000, 5000)

# Mix of positional + keyword
student_fee("Diana", "B.Sc", scholarship=2000, fee=12000)
````

````python
# ═══════════════════════════════════════════════════
# PROGRAM 3.2 — *args and **kwargs (Variable Arguments)
# ═══════════════════════════════════════════════════

# *args — accepts any number of positional arguments
def total_marks(*args):
    """Calculates total and average of any number of subjects."""
    total = sum(args)
    avg   = total / len(args)
    print(f"  Marks  : {args}")
    print(f"  Total  : {total}  |  Average : {round(avg, 2)}")

print("\n📊 Student 1 — 3 subjects:")
total_marks(78, 85, 90)

print("\n📊 Student 2 — 5 subjects:")
total_marks(65, 70, 88, 92, 74)

# **kwargs — accepts any number of keyword arguments
def student_profile(**kwargs):
    """Displays student profile from keyword arguments."""
    print("\n  📋 Student Profile:")
    for key, value in kwargs.items():
        print(f"     {key.capitalize():<12}: {value}")

student_profile(name="Alice", age=20, branch="CSE", cgpa=9.1, city="Chennai")
````

## 📌 Topic 4: The Return Statement

### What does `return` do?
The `return` statement:
1. **Sends a value back** to the caller
2. **Exits the function** immediately (any code after `return` is not executed)
3. A function **without** `return` automatically returns `None`

### Returning Multiple Values:
Python allows returning multiple values as a **tuple**:
```python
def min_max(lst):
    return min(lst), max(lst)   # returns a tuple

lo, hi = min_max([3, 1, 5, 2])  # unpacking
```

### Key Rules:
- `return` with no value → returns `None`
- Multiple `return` statements are allowed (useful with `if-else`)
- The returned value can be stored, printed, or passed directly

---


````python
# ═══════════════════════════════════════════════════
# PROGRAM 4.1 — Multiple Return Values
# ═══════════════════════════════════════════════════

def analyze_list(numbers):
    """Returns multiple statistical values from a list."""
    total   = sum(numbers)
    count   = len(numbers)
    average = round(total / count, 2)
    minimum = min(numbers)
    maximum = max(numbers)
    spread  = maximum - minimum
    return total, count, average, minimum, maximum, spread

# Unpack all returned values
scores = [72, 88, 65, 91, 78, 54, 95, 83]
t, c, avg, lo, hi, sp = analyze_list(scores)

print("\n📊 Statistical Analysis")
print("=" * 35)
print(f"  Scores   : {scores}")
print(f"  Count    : {c}")
print(f"  Total    : {t}")
print(f"  Average  : {avg}")
print(f"  Minimum  : {lo}")
print(f"  Maximum  : {hi}")
print(f"  Spread   : {sp}")
````

````python
# ═══════════════════════════════════════════════════
# PROGRAM 4.2 — Early Return & Return None Demonstration
# ═══════════════════════════════════════════════════

def safe_divide(a, b):
    """Divides a by b. Returns None if division by zero."""
    if b == 0:
        print("  ⚠️  Cannot divide by zero!")
        return None          # early exit
    return a / b             # normal return

def classify_temperature(temp):
    """Returns a label based on temperature in Celsius."""
    if temp < 0:
        return "Freezing 🧊"
    elif temp < 15:
        return "Cold ❄️"
    elif temp < 25:
        return "Pleasant 🌤️"
    elif temp < 35:
        return "Warm ☀️"
    else:
        return "Hot 🔥"

# Test safe_divide
print("Division Results:")
for a, b in [(10, 2), (7, 0), (15, 4)]:
    result = safe_divide(a, b)
    print(f"  {a} ÷ {b} = {result}")

# Test classify_temperature
temps = [-5, 10, 22, 31, 42]
print("\nTemperature Classification:")
for t in temps:
    print(f"  {t}°C → {classify_temperature(t)}")
````

## 📌 Topic 5: Using Branching, Looping and Data Structures in Functions

### Concept:
Functions become powerful when combined with:
- **Branching (`if-elif-else`)** — for decision making inside functions
- **Loops (`for`, `while`)** — for repetitive operations
- **Data Structures (lists, dicts, sets, tuples)** — for organized data handling

### This is where real-world problem solving happens!
Functions act as **mini-programs** — they can take a list, loop through it, make decisions, and return results.

---


````python
# ═══════════════════════════════════════════════════
# PROGRAM 5.1 — Grade Report System (Branching + Loops + Lists)
# ═══════════════════════════════════════════════════

def assign_grade(score):
    """Returns letter grade for a given score."""
    if score >= 90:   return "O  (Outstanding)"
    elif score >= 75: return "A  (Excellent)"
    elif score >= 60: return "B  (Good)"
    elif score >= 50: return "C  (Average)"
    elif score >= 35: return "D  (Pass)"
    else:             return "F  (Fail)"

def generate_report(student_name, subject_scores: dict):
    """
    Takes a dictionary of {subject: score} and prints a full
    grade report using loops and branching.
    """
    print(f"\n{'='*50}")
    print(f"  GRADE REPORT  —  {student_name}")
    print(f"{'='*50}")
    print(f"  {'Subject':<18} | {'Score':>5} | Grade")
    print(f"  {'-'*44}")

    total, count, fail_count = 0, 0, 0
    for subject, score in subject_scores.items():
        grade = assign_grade(score)
        print(f"  {subject:<18} | {score:>5} | {grade}")
        total += score
        count += 1
        if score < 35:
            fail_count += 1

    avg = round(total / count, 2)
    print(f"  {'-'*44}")
    print(f"  {'Average':<18} | {avg:>5} |")
    print(f"  Result: {'❌ FAIL' if fail_count > 0 else '✅ PASS'}")
    print(f"{'='*50}")

# Test
alice_marks = {"Mathematics": 88, "Physics": 72, "Chemistry": 65,
               "English": 90, "CS": 95}
generate_report("Alice", alice_marks)

bob_marks = {"Mathematics": 45, "Physics": 30, "Chemistry": 55,
             "English": 60, "CS": 70}
generate_report("Bob", bob_marks)
````

````python
# ═══════════════════════════════════════════════════
# PROGRAM 5.2 — Inventory Management (Dicts + Loops + Branching)
# ═══════════════════════════════════════════════════

def add_item(inventory, name, qty, price):
    inventory[name] = {"quantity": qty, "price": price}

def sell_item(inventory, name, qty_sold):
    if name not in inventory:
        return f"❌ '{name}' not found in inventory."
    if inventory[name]["quantity"] < qty_sold:
        return f"⚠️  Insufficient stock for '{name}'."
    inventory[name]["quantity"] -= qty_sold
    revenue = qty_sold * inventory[name]["price"]
    return f"✅ Sold {qty_sold} x {name}. Revenue: ₹{revenue}"

def low_stock_alert(inventory, threshold=5):
    """Returns list of items below threshold quantity."""
    alerts = [name for name, data in inventory.items()
              if data["quantity"] <= threshold]
    return alerts

def display_inventory(inventory):
    print(f"\n{'Item':<15} | {'Qty':>5} | {'Price':>8} | {'Value':>10}")
    print("-" * 48)
    total_value = 0
    for name, data in inventory.items():
        value = data["quantity"] * data["price"]
        total_value += value
        print(f"{name:<15} | {data['quantity']:>5} | ₹{data['price']:>7} | ₹{value:>9}")
    print("-" * 48)
    print(f"{'Total Stock Value':<15}                    ₹{total_value:>9}")

# Build inventory
stock = {}
add_item(stock, "Pen",      50,  10)
add_item(stock, "Notebook",  8, 120)
add_item(stock, "Eraser",    3,   5)
add_item(stock, "Ruler",    20,  30)
add_item(stock, "Compass",   4,  85)

display_inventory(stock)

print("\n--- Transactions ---")
print(sell_item(stock, "Pen",      15))
print(sell_item(stock, "Notebook", 10))
print(sell_item(stock, "Ruler",     5))

alerts = low_stock_alert(stock)
print(f"\n🔔 Low Stock Alert: {alerts}")
````

````python
# ═══════════════════════════════════════════════════
# PROGRAM 5.3 — Prime Numbers, Fibonacci using Functions + Loops
# ═══════════════════════════════════════════════════

def is_prime(n):
    """Returns True if n is a prime number."""
    if n < 2:
        return False
    for i in range(2, int(n**0.5) + 1):
        if n % i == 0:
            return False
    return True

def primes_in_range(start, end):
    """Returns list of primes between start and end."""
    return [n for n in range(start, end + 1) if is_prime(n)]

def fibonacci_series(count):
    """Generates Fibonacci series up to 'count' terms."""
    series = []
    a, b = 0, 1
    for _ in range(count):
        series.append(a)
        a, b = b, a + b
    return series

# Test prime function
print("Primes between 1 and 50:")
print(primes_in_range(1, 50))

print("\nPrimes between 100 and 130:")
print(primes_in_range(100, 130))

print("\nFirst 12 Fibonacci Numbers:")
fib = fibonacci_series(12)
print(fib)
print(f"Sum of first 12 Fibonacci: {sum(fib)}")
````

## 📌 Topic 6: Recursion

### What is Recursion?
**Recursion** is a technique where a **function calls itself** to solve a smaller version of the same problem.

### Structure of a Recursive Function:
```python
def recursive_function(problem):
    if base_case:            # ← STOP condition (MUST have this!)
        return result
    else:
        return recursive_function(smaller_problem)  # ← calls itself
```

### Two Essential Parts:
| Part | Description |
|------|-------------|
| **Base Case** | The stopping condition — prevents infinite recursion |
| **Recursive Case** | The function calls itself with a simpler/smaller input |

### Internal Workflow of Recursion (Call Stack):
When `factorial(4)` is called:
```
factorial(4)
  └── 4 * factorial(3)
           └── 3 * factorial(2)
                    └── 2 * factorial(1)
                             └── 1 * factorial(0)
                                      └── return 1   ← BASE CASE
                             └── 1 * 1 = 1
                    └── 2 * 1 = 2
           └── 3 * 2 = 6
  └── 4 * 6 = 24   ← Final Answer
```
Each call **waits** for the next to return → this is the **call stack**.

### ⚠️ Recursion Depth:
Python limits recursion to **1000 calls** by default (to prevent stack overflow).  
Use `sys.setrecursionlimit(n)` to increase it (with caution).

---


````python
# ═══════════════════════════════════════════════════
# PROGRAM 6.1 — Factorial: Recursive vs Iterative
# ═══════════════════════════════════════════════════

# Recursive version
def factorial_recursive(n):
    """Recursive factorial — calls itself until n reaches 0."""
    if n == 0 or n == 1:      # Base Case
        return 1
    return n * factorial_recursive(n - 1)   # Recursive Case

# Iterative version (for comparison)
def factorial_iterative(n):
    """Iterative factorial using a loop."""
    result = 1
    for i in range(2, n + 1):
        result *= i
    return result

# Trace recursive calls step by step
def factorial_trace(n, depth=0):
    indent = "  " * depth
    print(f"{indent}→ factorial({n}) called")
    if n == 0 or n == 1:
        print(f"{indent}← Base case reached: return 1")
        return 1
    result = n * factorial_trace(n - 1, depth + 1)
    print(f"{indent}← factorial({n}) = {n} × ... = {result}")
    return result

print("🔍 Recursive Call Trace for factorial(5):")
print("-" * 42)
ans = factorial_trace(5)
print(f"\n✅ Final Answer: {ans}")

print("\n📊 Comparison Table:")
print(f"{'n':<5} | {'Recursive':>12} | {'Iterative':>12}")
print("-" * 35)
for n in range(0, 11):
    print(f"{n:<5} | {factorial_recursive(n):>12} | {factorial_iterative(n):>12}")
````

````python
# ═══════════════════════════════════════════════════
# PROGRAM 6.2 — Fibonacci (Recursive) with Memoization
# ═══════════════════════════════════════════════════

# Simple recursive (slow — recomputes same values)
def fib_recursive(n):
    if n <= 1:
        return n
    return fib_recursive(n-1) + fib_recursive(n-2)

# Memoized recursive (fast — stores computed values)
def fib_memo(n, memo={}):
    """Fibonacci with memoization to avoid redundant calls."""
    if n in memo:
        return memo[n]
    if n <= 1:
        return n
    memo[n] = fib_memo(n-1, memo) + fib_memo(n-2, memo)
    return memo[n]

import time

print("Fibonacci Series (Recursive):")
print([fib_recursive(n) for n in range(15)])

print("\nFibonacci Series (Memoized):")
print([fib_memo(n) for n in range(20)])

# Speed comparison
print("\n⏱️  Speed Comparison (fib(30)):")
t1 = time.time()
r1 = fib_recursive(30)
t2 = time.time()
r2 = fib_memo(30, {})
t3 = time.time()
print(f"  Simple Recursive : {r1} in {round((t2-t1)*1000, 3)} ms")
print(f"  Memoized         : {r2} in {round((t3-t2)*1000, 3)} ms")
````

````python
# ═══════════════════════════════════════════════════
# PROGRAM 6.3 — Tower of Hanoi (Classic Recursion)
# ═══════════════════════════════════════════════════

"""
Tower of Hanoi:
Move n disks from source peg to destination peg using an auxiliary peg.
Rules:
  1. Only one disk can be moved at a time.
  2. A disk can only be placed on a larger disk.
  3. Only the top disk of any peg can be moved.
"""

move_count = 0

def hanoi(n, source, destination, auxiliary):
    global move_count
    if n == 1:
        move_count += 1
        print(f"  Step {move_count:>2}: Move disk 1  from  '{source}'  →  '{destination}'")
        return
    hanoi(n-1, source, auxiliary, destination)   # move n-1 disks to aux
    move_count += 1
    print(f"  Step {move_count:>2}: Move disk {n}  from  '{source}'  →  '{destination}'")
    hanoi(n-1, auxiliary, destination, source)   # move n-1 disks to dest

for disks in [2, 3]:
    move_count = 0
    print(f"\n🗼 Tower of Hanoi — {disks} Disks:")
    print("-" * 48)
    hanoi(disks, "A (Source)", "C (Dest)", "B (Aux)")
    print(f"  ✅ Total moves required: {move_count}  (Formula: 2^n - 1 = {2**disks - 1})")
````

````python
# ═══════════════════════════════════════════════════
# PROGRAM 6.4 — Recursive Sum, Power and Palindrome Check
# ═══════════════════════════════════════════════════

def recursive_sum(lst):
    """Returns sum of list elements using recursion."""
    if len(lst) == 0:           # Base: empty list
        return 0
    return lst[0] + recursive_sum(lst[1:])  # head + sum of tail

def power(base, exp):
    """Returns base^exp using recursion."""
    if exp == 0:                # Any number^0 = 1
        return 1
    return base * power(base, exp - 1)

def is_palindrome(s):
    """Checks if string is palindrome using recursion."""
    s = s.lower().replace(" ", "")
    if len(s) <= 1:             # Base: 0 or 1 char
        return True
    if s[0] != s[-1]:           # First and last don't match
        return False
    return is_palindrome(s[1:-1])  # Check inner string

# Test recursive_sum
nums = [10, 20, 30, 40, 50]
print(f"Recursive Sum of {nums} = {recursive_sum(nums)}")

# Test power
print("\nPower using Recursion:")
for b, e in [(2,10), (3,5), (5,3), (10,4)]:
    print(f"  {b}^{e} = {power(b,e)}")

# Test palindrome
words = ["racecar", "hello", "madam", "python",
         "A man a plan a canal Panama", "level"]
print("\nPalindrome Check:")
for w in words:
    result = "✅ Palindrome" if is_palindrome(w) else "❌ Not Palindrome"
    print(f"  '{w}' → {result}")
````

## 📌 Topic 7: Modules in Python

### What is a Module?
A **module** is a file containing Python code (functions, classes, variables) that can be **imported and reused** in other programs.

### Types of Modules:
| Type | Description | Example |
|------|-------------|---------|
| **Built-in Modules** | Come with Python | `math`, `random`, `os`, `sys` |
| **Third-party Modules** | Installed via pip | `numpy`, `pandas`, `requests` |

````python
# ═══════════════════════════════════════════════════
# PROGRAM 7.2 — The random Module: OTP + Dice + Lottery
# ═══════════════════════════════════════════════════

import random

# 1. Generate a 6-digit OTP
otp = ''.join([str(random.randint(0, 9)) for _ in range(6)])
print(f"🔐 Generated OTP       : {otp}")

# 2. Simulate rolling two dice
d1, d2 = random.randint(1, 6), random.randint(1, 6)
print(f"🎲 Dice Roll           : {d1} + {d2} = {d1 + d2}")

# 3. Shuffle a deck (simplified)
deck = ["A♠", "K♠", "Q♠", "J♠", "10♠", "A♥", "K♥", "Q♥"]
random.shuffle(deck)
print(f"🃏 Shuffled Cards      : {deck[:5]} ...")

# 4. Pick lottery winners from class
students = ["Alice", "Bob", "Charlie", "Diana", "Eve",
            "Frank", "Grace", "Heidi"]
winners = random.sample(students, 3)
print(f"🏆 Lottery Winners     : {winners}")

# 5. Random float in range
price_fluctuation = random.uniform(-5.0, 5.0)
print(f"📈 Price Change (%)    : {round(price_fluctuation, 2)}%")

# 6. Seed for reproducibility
random.seed(42)
print(f"🔁 Seeded random list  : {[random.randint(1,100) for _ in range(5)]}")
````

````python
# ═══════════════════════════════════════════════════
# PROGRAM 7.3 — Creating and Using a User-Defined Module
# (Simulated — module code written as string for demonstration)
# ═══════════════════════════════════════════════════

# ─── This is what 'mathutils.py' would contain ───
module_code = '''
"""mathutils.py — A custom math utilities module."""

PI = 3.14159265358979

def circle_area(radius):
    return round(PI * radius ** 2, 4)

def circle_perimeter(radius):
    return round(2 * PI * radius, 4)

def rectangle_area(length, breadth):
    return length * breadth

def rectangle_perimeter(length, breadth):
    return 2 * (length + breadth)

def is_perfect_square(n):
    import math
    root = int(math.sqrt(n))
    return root * root == n
'''

print("📄 Contents of 'mathutils.py':")
print("-" * 40)
print(module_code)

# ─── Simulating the module inline ───
exec(module_code)

print("─" * 40)
print("Using mathutils functions:")
print(f"  Circle area (r=7)         = {circle_area(7)}")
print(f"  Circle perimeter (r=7)    = {circle_perimeter(7)}")
print(f"  Rectangle area (5×8)      = {rectangle_area(5, 8)}")
print(f"  Rectangle perimeter (5×8) = {rectangle_perimeter(5, 8)}")
print(f"  Is 144 perfect square?    = {is_perfect_square(144)}")
print(f"  Is 50 perfect square?     = {is_perfect_square(50)}")
````

````python
# ═══════════════════════════════════════════════════
# PROGRAM 7.4 — datetime and statistics Modules
# ═══════════════════════════════════════════════════

from datetime import datetime, date, timedelta
import statistics

# ─── datetime ───
now = datetime.now()
print("📅 datetime Module:")
print(f"  Current date & time : {now.strftime('%d %B %Y, %I:%M %p')}")
print(f"  Current year        : {now.year}")
print(f"  Day of week         : {now.strftime('%A')}")

# Age calculator
dob = date(2003, 8, 15)
today = date.today()
age = today.year - dob.year - ((today.month, today.day) < (dob.month, dob.day))
print(f"  Age (born {dob})  : {age} years")

# Deadline countdown
exam = date(2025, 12, 1)
days_left = (exam - today).days
print(f"  Days until exam     : {max(days_left, 0)} days")

# ─── statistics ───
scores = [72, 85, 90, 68, 74, 85, 91, 77, 83, 85]
print("\n📊 statistics Module:")
print(f"  Data     : {scores}")
print(f"  Mean     : {statistics.mean(scores)}")
print(f"  Median   : {statistics.median(scores)}")
print(f"  Mode     : {statistics.mode(scores)}")
print(f"  Std Dev  : {round(statistics.stdev(scores), 3)}")
print(f"  Variance : {round(statistics.variance(scores), 3)}")
````

| Topic | Key Concept |
|-------|-------------|
| **Pre-defined Functions** | Ready-to-use functions built into Python (`len`, `sum`, `max`, etc.) |

| **Call Stack** | Internal memory stack Python uses to track recursive calls |
| **Memoization** | Optimization technique for recursion — cache computed results |
| **Modules** | Python files with reusable code; import using `import` or `from` |
