## 1. 📄 What is a File?

A **file** is a named location on disk that stores related data permanently.

| Type | Description | Example |
|------|-------------|-----------------------------------------------------------|
| Text File | Stores data as human-readable characters | `.txt`, `.csv`, `.py` |
| Binary File | Stores data as raw bytes (not human-readable) | `.jpg`, `.mp3`, `.exe` |

Python provides **built-in functions** to handle files without importing extra libraries (for basic operations).

## 2. 🔓 Opening a File — `open()` Function

### Syntax
```python
file_object = open(filename, mode, encoding='utf-8')
```

- `filename` — Name or path of the file
- `mode` — How to open the file (read, write, append, etc.)

| JSON files | UTF-8 |

| Multilingual data | UTF-8 |

## 3. 🗂️ File Modes

| Mode | Symbol | Description |
|------|--------|--------------------------------------------------------------|
| Read | `'r'` | Opens file for reading. File must exist. (default) |
| Write | `'w'` | Opens for writing. Creates new file or **erases existing**. |
| Append | `'a'` | Opens for appending. Does NOT erase existing content. |
| Read+Write | `'r+'` | Opens for both reading and writing. File must exist. |
| Write+Read | `'w+'` | Read and write. Erases existing content. |
| Binary Read | `'rb'` | Read in binary mode. |
| Binary Write | `'wb'` | Write in binary mode. |
| Exclusive Create | `'x'` | Creates file; **fails if file already exists**. |

````python
# Demo: Understanding modes
import os

# 'w' mode — creates or overwrites
with open('mode_demo.txt', 'w') as f:
    f.write('Line 1\n')
    f.write('Line 2\n')

# 'x' mode — creates only if file does NOT exist
if not os.path.exists('new_file.txt'):
    with open('new_file.txt', 'x') as f:
        f.write('Created exclusively!')
    print('new_file.txt created.')
else:
    print('new_file.txt already exists — x mode would raise FileExistsError!')
````

> ⚠️ **Care to take:** Using `'w'` mode on an existing file **permanently deletes** all its previous content without warning. Use `'a'` if you want to preserve existing data, or check existence before writing.

````python
# Method 1: read() — Entire file at once
with open('fruits.txt', 'r', encoding='utf-8') as f:
    content = f.read()
print('--- read() ---')
print(content)
````

````python
# Method 2: read(n) — Read first n characters
with open('fruits.txt', 'r', encoding='utf-8') as f:
    partial = f.read(5)   # Read only first 5 characters
print('--- read(5) ---')
print(partial)   # Output: Apple
````

````python
# Method 3: readline() — One line at a time
with open('fruits.txt', 'r', encoding='utf-8') as f:
    line1 = f.readline()
    line2 = f.readline()
print('--- readline() ---')
print('Line 1:', line1.strip())
print('Line 2:', line2.strip())
````

````python
# Method 4: readlines() — Returns a list of all lines
with open('fruits.txt', 'r', encoding='utf-8') as f:
    lines = f.readlines()
print('--- readlines() ---')
print(lines)
print('Total lines:', len(lines))
````

````python
# Method 5: Iterating line by line (✅ Best for large files)
print('--- for loop ---')
with open('fruits.txt', 'r', encoding='utf-8') as f:
    for line in f:
        print(line.strip())   # strip() removes trailing \n
````

> ⚠️ **Care to take:**
> - `read()` and `readlines()` load the **entire file into memory** — avoid on large files (use `for` loop instead).
> - Each `readline()` call moves the file pointer forward — reading continues from where you left off.
> - Always call `.strip()` on lines to remove trailing newline `\n` characters when processing data.

## 5. ✏️ Writing to a File

| Method | Description |
|-----------|---------------------------------------------|
| `write(s)` | Writes a string to the file |
| `writelines(list)` | Writes a list of strings to the file |

````python
# write() — Write a single string
with open('output.txt', 'w', encoding='utf-8') as f:
    f.write('Name: Venkatesh\n')
    f.write('Course: Problem Solving Using Python\n')
    f.write('Grade: A\n')

print('File written. Reading back...')
with open('output.txt', 'r') as f:
    print(f.read())
````

````python
# writelines() — Write a list of strings
# NOTE: writelines() does NOT add newlines automatically!
students = ['Alice\n', 'Bob\n', 'Charlie\n', 'Diana\n']

with open('students.txt', 'w', encoding='utf-8') as f:
    f.writelines(students)

print('Students file written. Reading back...')
with open('students.txt', 'r') as f:
    print(f.read())
````

> ⚠️ **Care to take:**
> - `write()` does **NOT automatically add newlines** — you must include `\n` yourself.
> - `writelines()` also does **NOT add newlines** between items — include `\n` in each string.
> - Using `'w'` mode on an existing file **silently overwrites** all previous content.

## 6. ➕ Appending to a File

Use `'a'` mode to **add content** to the end of an existing file without deleting previous content.

````python
# Append new students to the existing students.txt
new_students = ['Eve\n', 'Frank\n']

with open('students.txt', 'a', encoding='utf-8') as f:
    f.writelines(new_students)

print('After appending:')
with open('students.txt', 'r') as f:
    print(f.read())
````

> ⚠️ **Care to take:**
> - If the file does **not exist**, `'a'` mode **creates it** — no error is raised.
> - If it **does exist**, data is always added at the **end**, not the beginning or middle.
> - To insert content in the middle of a file, you must read, modify, and rewrite the entire file.

## 7. 🛡️ The `with` Statement (Context Manager)

The `with` statement is the **recommended way** to work with files.

- Automatically **closes the file** when the block ends — even if an error occurs.
- Cleaner and safer than manually calling `close()`.

```python
with open('file.txt', 'r') as f:
    data = f.read()
# File is automatically closed here
```

````python
# Without 'with' — risky
f = open('output.txt', 'r')
data = f.read()
f.close()  # What if an error occurred before this line? File stays open!
print('Without with:', data[:30])

# ✅ With 'with' — safe and clean
with open('output.txt', 'r') as f:
    data = f.read()
# File is closed automatically here
print('With statement:', data[:30])
print('Is file closed?', f.closed)  # True
````

````python
# Opening multiple files at once using 'with'
with open('fruits.txt', 'r') as src, open('fruits_copy.txt', 'w') as dst:
    for line in src:
        dst.write(line)

print('File copied successfully!')
with open('fruits_copy.txt') as f:
    print(f.read())
````

## 8. 🔢 Working with Binary Files

Binary mode (`'rb'`, `'wb'`) is used for **non-text files** like images, audio, PDFs, etc.

````python
# Writing binary data
data = bytes([72, 101, 108, 108, 111])  # ASCII codes for 'Hello'

with open('binary_file.bin', 'wb') as f:
    f.write(data)

# Reading binary data
with open('binary_file.bin', 'rb') as f:
    content = f.read()

print('Raw bytes:', content)
print('Decoded text:', content.decode('utf-8'))
````

````python
# Copying a binary file (image, pdf, etc.) — Chunk-by-chunk for large files
def copy_binary_file(src_path, dst_path, chunk_size=1024):
    with open(src_path, 'rb') as src, open(dst_path, 'wb') as dst:
        while True:
            chunk = src.read(chunk_size)
            if not chunk:
                break
            dst.write(chunk)
    print(f'Copied {src_path} → {dst_path}')

copy_binary_file('binary_file.bin', 'binary_copy.bin')
````

> ⚠️ **Care to take:**
> - Never open a binary file in **text mode** (`'r'` or `'w'`) — it will corrupt the data.
> - On Windows, text mode converts `\n` to `\r\n`, which can corrupt binary files.
> - Always specify encoding when opening **text** files; always use `'b'` flag for **binary** files.

## 9. 📍 File Pointer — `seek()` and `tell()`

The **file pointer** tracks the current read/write position in the file.

| Method | Description |
|--------|-------------------------------------------|
| `tell()` | Returns current position (byte offset) |
| `seek(offset, from)` | Moves pointer to given position |

````python
with open('fruits.txt', 'r', encoding='utf-8') as f:
    print('Initial position:', f.tell())       # 0
    
    first = f.read(5)
    print(f'After read(5): "{first}"')
    print('Position after read(5):', f.tell()) # 5
    
    f.seek(0)                                   # Go back to beginning
    print('After seek(0):', f.tell())           # 0
    
    again = f.read(5)
    print(f'Reading again: "{again}"')          # Apple (same as first)
````

````python
# Seek reference points:
# seek(offset, 0) — from beginning (default)
# seek(offset, 1) — from current position
# seek(offset, 2) — from end of file

with open('fruits.txt', 'rb') as f:   # Binary mode needed for seek from end
    f.seek(0, 2)                       # Move to end
    size = f.tell()
    print(f'File size: {size} bytes')
    
    f.seek(-7, 2)                      # 7 bytes before end
    tail = f.read()
    print('Last 7 bytes:', tail)
````

## 11. 🔴 Exception Handling in File I/O

File operations can fail for many reasons:
- File not found
- No read/write permission
- Disk is full
- File is already open by another process

Always handle exceptions!

| Exception | When it occurs |
|-----------|----------------|
| `FileNotFoundError` | File doesn't exist (mode `'r'`) |
| `PermissionError` | No permission to read/write |
| `FileExistsError` | File already exists (mode `'x'`) |
| `IsADirectoryError` | Tried to open a directory as a file |
| `IOError` / `OSError` | General I/O error |

````python
# ✅ Safe file reading with exception handling
def safe_read(filepath):
    try:
        with open(filepath, 'r', encoding='utf-8') as f:
            return f.read()
    except FileNotFoundError:
        print(f'❌ Error: File "{filepath}" not found.')
    except PermissionError:
        print(f'❌ Error: No permission to read "{filepath}".')
    except OSError as e:
        print(f'❌ OS Error: {e}')
    return None

# Test with existing file
content = safe_read('fruits.txt')
print('Content:', content)

# Test with missing file
safe_read('missing_file.txt')
````

````python
# ✅ Safe file writing with exception handling
def safe_write(filepath, data):
    try:
        with open(filepath, 'w', encoding='utf-8') as f:
            f.write(data)
        print(f'✅ Successfully written to "{filepath}"')
    except PermissionError:
        print(f'❌ No permission to write to "{filepath}"')
    except OSError as e:
        print(f'❌ Error writing file: {e}')

safe_write('result.txt', 'Final exam result: Pass!')
````

## 12. ⚠️ Common Mistakes & Care to be Taken

This section summarises all important precautions in one place.

````python
# ❌ MISTAKE 1: Forgetting to close the file
f = open('output.txt', 'r')
data = f.read()
# If an error occurs here, f.close() is never called!
f.close()

# ✅ FIX: Use 'with' statement
with open('output.txt', 'r') as f:
    data = f.read()
# Automatically closed
````

````python
# ❌ MISTAKE 2: Overwriting important files with 'w' mode
# with open('important_data.txt', 'w') as f:  # Erases everything!
#     f.write('new stuff')

# ✅ FIX: Check before writing or use 'a' to append
import os
filepath = 'important_data.txt'
if os.path.exists(filepath):
    print(f'⚠️ Warning: {filepath} already exists! Not overwriting.')
else:
    with open(filepath, 'w') as f:
        f.write('safe to write now')
    print('Written safely.')
````

````python
# ❌ MISTAKE 3: Trailing newlines not stripped
with open('fruits.txt', 'r') as f:
    lines = f.readlines()

print('Without strip:', repr(lines[0]))   # 'Apple\n'

# ✅ FIX: Use strip()
clean_lines = [line.strip() for line in lines]
print('With strip:   ', repr(clean_lines[0]))  # 'Apple'
````

````python
# ❌ MISTAKE 4: Not specifying encoding
# with open('file.txt', 'r') as f:  # Uses system default — may fail on different OS

# ✅ FIX: Always specify encoding='utf-8' for portability
with open('fruits.txt', 'r', encoding='utf-8') as f:
    content = f.read()
print('Encoding-safe read done.')
````

````python
# ❌ MISTAKE 5: Reading a closed file
with open('fruits.txt', 'r') as f:
    pass  # Do nothing inside — file closes at end of 'with'

try:
    data = f.read()  # f is now CLOSED
except ValueError as e:
    print('❌ Error:', e)   # I/O operation on closed file

# ✅ FIX: Do all file operations inside the 'with' block
````

````python
# ❌ MISTAKE 6: Using write() with non-string data
with open('numbers.txt', 'w') as f:
    try:
        f.write(12345)  # ❌ TypeError: must be str, not int
    except TypeError as e:
        print('❌ Error:', e)

# ✅ FIX: Always convert to string first
with open('numbers.txt', 'w') as f:
    f.write(str(12345))  # ✅
    f.write('\n')
    f.write(str(3.14))
print('Numbers written to file.')
````

## 13. 📋 Quick Reference Summary

### File Open Modes
| Mode | Meaning | File must exist? | Overwrites? |
|------|---------|-----------------|-------------|
| `r` | Read | Yes | No |
| `w` | Write | No (creates) | Yes |
| `a` | Append | No (creates) | No |
| `x` | Exclusive create | No (must NOT exist) | — |
| `r+` | Read + Write | Yes | No |
| `rb/wb` | Binary modes | Same as above | Same |

### Essential Best Practices

| # | Rule |
|---|------|
| 1 | ✅ Always use `with` statement |
| 2 | ✅ Specify `encoding='utf-8'` for text files |
| 3 | ✅ Use `strip()` when reading lines |
| 4 | ✅ Convert data to `str` before writing |
| 5 | ✅ Check `os.path.exists()` before delete/rename |
| 6 | ✅ Use `os.path.join()` for cross-platform paths |
| 7 | ✅ Use `try-except` for all file operations |
| 8 | ✅ Use `'rb'/'wb'` for binary files (images, PDFs) |
| 9 | ⚠️ Never use `'w'` on important files without backup |
| 10 | ⚠️ Avoid `read()` on very large files — iterate instead |
