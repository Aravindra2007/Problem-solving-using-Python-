## 📌 Section 1: Numeric Types in Python

Python supports four built-in numeric types:

| Type | Description | Example |
|------|-------------|--------|
| `int` | Whole numbers (unlimited precision) | `42`, `-7`, `0` |
| `float` | Decimal numbers (64-bit IEEE 754) | `3.14`, `-0.001` |
| `complex` | Real + Imaginary parts | `3+4j` |
| `bool` | Subclass of int; True=1, False=0 | `True` |

**Key points:**
- Python integers have **arbitrary precision** — no overflow ever.
- Floats follow IEEE 754 double precision and can have tiny rounding errors.
- Complex numbers use `j` (not `i`) for the imaginary unit.
- `bool` is a subclass of `int`: `True == 1` and `False == 0`.


### Program 1 – Exploring int, float, complex, bool

````python
# Program 1: Numeric Types
# ─────────────────────────────────────────────────────────────

# int: multiple literal bases
a     = 1_000_000   # underscores for readability (Python 3.6+)
b_bin = 0b1010      # binary  → 10
b_oct = 0o17        # octal   → 15
b_hex = 0xFF        # hex     → 255

print('── int ───────────────────────────────────────────')
print(f'  1_000_000 = {a}')
print(f'  0b1010    = {b_bin}   (binary)')
print(f'  0o17      = {b_oct}   (octal)')
print(f'  0xFF      = {b_hex}  (hex)')
print(f'  type(a)   = {type(a).__name__}')

# float
pi  = 3.141592653589793
sci = 1.5e-4
print('\n── float ─────────────────────────────────────────')
print(f'  pi      = {pi}')
print(f'  1.5e-4  = {sci}  =>  {sci:.10f}')

# complex
z = 3 + 4j
print('\n── complex ───────────────────────────────────────')
print(f'  z         = {z}')
print(f'  real part = {z.real}')
print(f'  imag part = {z.imag}')
print(f'  |z|       = {abs(z)}')

# bool (subclass of int)
print('\n── bool ──────────────────────────────────────────')
print(f'  True + True       = {True + True}')
print(f'  True * 5          = {True * 5}')
print(f'  False + 100       = {False + 100}')
print(f'  isinstance(True, int) = {isinstance(True, int)}')
````

### Program 3 – Bitwise Operators on Integers

````python
# Program 3: Bitwise Operations
# ─────────────────────────────────────────────────────────────

a, b = 0b1100, 0b1010   # 12 and 10
print(f'a = {a}  =>  {bin(a)}')
print(f'b = {b}  =>  {bin(b)}')
print()
print(f'a & b  (AND)       = {a & b:3d}  =>  {bin(a & b)}')
print(f'a | b  (OR)        = {a | b:3d}  =>  {bin(a | b)}')
print(f'a ^ b  (XOR)       = {a ^ b:3d}  =>  {bin(a ^ b)}')
print(f'~a     (NOT)       = {~a:3d}  (two\'s complement)')
print(f'a << 1 (left  sh)  = {a << 1:3d}  =>  {bin(a << 1)}')
print(f'a >> 1 (right sh)  = {a >> 1:3d}  =>  {bin(a >> 1)}')

# Even/Odd via bit mask
print('\nEven/Odd via bit masking:')
for n in range(8):
    parity = 'even' if (n & 1) == 0 else 'odd'
    print(f'  {n}: {parity}')

# XOR swap
x, y = 42, 99
x ^= y; y ^= x; x ^= y
print(f'\nXOR swap: x={x}, y={y}')
````

### Program 5 – Base Conversion Utilities

````python
# Program 5: Number Base Conversions
# ─────────────────────────────────────────────────────────────

n = 255
print(f'Decimal : {n}')
print(f'Binary  : {bin(n)}  =>  stripped: {bin(n)[2:]}')
print(f'Octal   : {oct(n)}  =>  stripped: {oct(n)[2:]}')
print(f'Hex     : {hex(n)}  =>  upper: {hex(n)[2:].upper()}')

def to_base(num, base):
    digits = '0123456789ABCDEFGHIJKLMNOPQRSTUVWXYZ'
    if num == 0: return '0'
    result = ''
    while num:
        result = digits[num % base] + result
        num //= base
    return result

print('\nCustom base-conversion for 255:')
for base in [2, 8, 10, 16, 32, 36]:
    print(f'  base {base:2d} => {to_base(255, base)}')

print('\nTemperature conversions (Celsius => Fahrenheit):')
for c in [0, 20, 37, 100]:
    print(f'  {c}C = {c * 9/5 + 32:.1f}F')
````

### Program 8 – Quadratic Equation Solver

````python
import math

# Program 8: Quadratic Equation Solver   ax^2 + bx + c = 0
# ─────────────────────────────────────────────────────────────

def solve_quadratic(a, b, c):
    D = b*b - 4*a*c
    print(f'  Equation : {a}x^2 + ({b})x + ({c}) = 0')
    print(f'  Discriminant D = {D}')
    if D > 0:
        x1 = (-b + math.sqrt(D)) / (2*a)
        x2 = (-b - math.sqrt(D)) / (2*a)
        print(f'  Two real roots  : x1={x1:.4f}, x2={x2:.4f}')
    elif D == 0:
        x = -b / (2*a)
        print(f'  One repeated root: x = {x:.4f}')
    else:
        real = -b / (2*a)
        imag = math.sqrt(-D) / (2*a)
        print(f'  Complex roots   : {real:.4f} +/- {imag:.4f}j')
    print()

solve_quadratic(1, -5,  6)   # roots: 2 and 3
solve_quadratic(1, -2,  1)   # root:  1 (repeated)
solve_quadratic(1,  1,  1)   # complex roots
solve_quadratic(2, -4, -6)   # roots: 3 and -1

# Verify first case
a, b, c = 1, -5, 6
print('Verification: plugging x=2 and x=3 back in:')
for x in [2, 3]:
    print(f'  f({x}) = {a*x**2 + b*x + c}')
````

### Program 10 – Dice Roll Simulation & Probability

````python
import random
from collections import Counter

# Program 10: Dice Simulation
# ─────────────────────────────────────────────────────────────

random.seed(0)

def roll_dice(sides=6, rolls=10000):
    results = [random.randint(1, sides) for _ in range(rolls)]
    counts = Counter(results)
    expected = rolls / sides
    print(f'Rolling d{sides} x {rolls:,}  (expected per face: {expected:.0f})')
    for face in range(1, sides+1):
        bar = '#' * (counts[face] // 50)
        pct = counts[face] / rolls * 100
        print(f'  {face}: {bar:<30} {pct:.1f}%')

roll_dice(6)

print()
print('Two-dice sum distribution (10,000 rolls):')
sums = [random.randint(1,6) + random.randint(1,6) for _ in range(10000)]
cnt = Counter(sums)
for s in range(2, 13):
    bar = '#' * (cnt[s] // 30)
    print(f'  {s:2d}: {bar:<45} {cnt[s]/100:.1f}%')
````

### Program 11 – OTP & Password Generator

````python
import random, string

# Program 11: OTP & Password Generator
# ─────────────────────────────────────────────────────────────

def generate_otp(length=6):
    return ''.join(random.choices(string.digits, k=length))

def generate_password(length=16, symbols=True):
    chars = string.ascii_letters + string.digits
    if symbols:
        chars += string.punctuation
    pwd = [
        random.choice(string.ascii_uppercase),
        random.choice(string.ascii_lowercase),
        random.choice(string.digits),
    ]
    if symbols:
        pwd.append(random.choice(string.punctuation))
    pwd += random.choices(chars, k=length - len(pwd))
    random.shuffle(pwd)
    return ''.join(pwd)

def check_strength(pwd):
    has_upper = any(c.isupper() for c in pwd)
    has_lower = any(c.islower() for c in pwd)
    has_digit = any(c.isdigit() for c in pwd)
    has_sym   = any(c in string.punctuation for c in pwd)
    score = sum([has_upper, has_lower, has_digit, has_sym, len(pwd) >= 12])
    levels = {5: 'Strong', 4: 'Good', 3: 'Moderate', 2: 'Weak'}
    return levels.get(score, 'Very Weak')

print('── 6-digit OTPs ─────────────────────────────────')
for _ in range(6):
    print(f'  {generate_otp()}')

print('\n── Strong Passwords ─────────────────────────────')
for _ in range(5):
    pwd = generate_password(16)
    print(f'  {pwd}  [{check_strength(pwd)}]')

print('\n── Lottery (6 of 49) ────────────────────────────')
for draw in range(3):
    nums = sorted(random.sample(range(1, 50), 6))
    bonus = random.randint(1, 49)
    print(f'  Draw {draw+1}: {nums}  Bonus: {bonus}')
````

### Program 13 – String Slicing

````python
# Program 13: String Slicing   s[start : stop : step]
# ─────────────────────────────────────────────────────────────

s = 'Hello, World!'
print(f's = "{s}"  (len={len(s)})')
print()

print('Basic slices:')
print(f'  s[0:5]    = "{s[0:5]}"     first 5 chars')
print(f'  s[7:]     = "{s[7:]}"   from index 7 to end')
print(f'  s[:5]     = "{s[:5]}"   first 5 (start=0 default)')
print(f'  s[-6:]    = "{s[-6:]}"   last 6 chars')
print(f'  s[-6:-1]  = "{s[-6:-1]}"  negative slice')
print(f'  s[:]      = "{s[:]}"  full copy')

print('\nStep slices:')
print(f'  s[::2]    = "{s[::2]}"   every 2nd char')
print(f'  s[::3]    = "{s[::3]}"     every 3rd')
print(f'  s[::-1]   = "{s[::-1]}"  REVERSED')

# URL parsing via slicing
url = 'https://www.example.com/page?q=python'
print(f'\nURL: {url}')
protocol = url[:url.index('://')]
rest     = url[url.index('://')+3:]
domain   = rest[:rest.index('/')]
path     = rest[rest.index('/'):]
print(f'  protocol = {protocol}')
print(f'  domain   = {domain}')
print(f'  path     = {path}')

# Palindrome via reverse slice
words = ['racecar', 'python', 'level', 'hello', 'madam']
print('\nPalindrome check:')
for w in words:
    print(f'  "{w}" palindrome? {w == w[::-1]}')
````

### Program 17 – Classic String Programs

````python
# Program 17: Palindrome, Anagram, Caesar Cipher, Word Frequency
# ─────────────────────────────────────────────────────────────

# 1. Palindrome checker
def is_palindrome(s):
    cleaned = ''.join(c.lower() for c in s if c.isalnum())
    return cleaned == cleaned[::-1]

tests = ['racecar','Hello','A man a plan a canal Panama','Was it a car or a cat I saw']
print('── Palindromes ──────────────────────────────────')
for t in tests:
    display = t if len(t) <= 35 else t[:32] + '...'
    print(f'  "{display}" => {is_palindrome(t)}')

# 2. Anagram checker
def is_anagram(a, b):
    clean = lambda s: sorted(s.replace(' ','').lower())
    return clean(a) == clean(b)

pairs = [('listen','silent'),('hello','world'),('Astronomer','Moon starer')]
print('\n── Anagrams ──────────────────────────────────────')
for a, b in pairs:
    print(f'  "{a}" & "{b}" => {is_anagram(a, b)}')

# 3. Caesar cipher
def caesar(text, shift):
    result = []
    for ch in text:
        if ch.isalpha():
            base = ord('A') if ch.isupper() else ord('a')
            result.append(chr((ord(ch) - base + shift) % 26 + base))
        else:
            result.append(ch)
    return ''.join(result)

msg = 'Hello, Python!'
for shift in [3, 13]:
    enc = caesar(msg, shift)
    dec = caesar(enc, -shift)
    print(f'\n── Caesar shift={shift} ───────────────────────────────')
    print(f'  Original  : {msg}')
    print(f'  Encrypted : {enc}')
    print(f'  Decrypted : {dec}')

# 4. Word frequency counter
text = 'to be or not to be that is the question to be'
freq = {}
for word in text.split():
    freq[word] = freq.get(word, 0) + 1
print('\n── Word Frequency ───────────────────────────────')
for word, count in sorted(freq.items(), key=lambda x: -x[1]):
    print(f'  {word:<12}: {"*" * count} ({count})')
````

## 📌 Section 3: File I/O in Python

### `open()` Syntax
```python
file_obj = open(filename, mode='r', encoding=None)
```

### File Modes

| Mode | Behaviour |
|------|----------|
| `'r'` | Read — file must exist |
| `'w'` | Write — creates; **truncates** if exists |
| `'a'` | Append — writes at end; creates if needed |
| `'x'` | Exclusive create — fails if file exists |
| `'b'` | Binary (combine: `'rb'`, `'wb'`) |
| `'+'` | Read+Write (combine: `'r+'`, `'w+'`, `'a+'`) |

### File Object Attributes
| Attribute | Meaning |
|-----------|--------|
| `.name` | Path/filename |
| `.mode` | Mode string |
| `.closed` | `True` if closed |
| `.encoding` | Encoding (text mode) |

### Best Practice: Always use `with`
```python
with open('file.txt', 'r') as f:
    data = f.read()
# file is automatically closed here
```


### Program 18 – open(), File Attributes, Write & Read

````python
import os

# Program 18: Basic File I/O
# ─────────────────────────────────────────────────────────────

filepath = '/tmp/demo_io.txt'

# WRITE (mode 'w')
print('── Writing ──────────────────────────────────────')
with open(filepath, 'w', encoding='utf-8') as f:
    print(f'  name      = {f.name}')
    print(f'  mode      = {f.mode}')
    print(f'  encoding  = {f.encoding}')
    print(f'  closed?   = {f.closed}')
    lines = [
        'Python File I/O Demo\n',
        'Line 2: Hello, World!\n',
        'Line 3: File handling is easy.\n',
        'Line 4: Always close your files.\n',
        'Line 5: Use with-blocks for safety.\n',
    ]
    f.writelines(lines)
    print(f'  Written {sum(len(l) for l in lines)} characters')
print(f'  After with: closed? = {f.closed}')

# READ ALL AT ONCE
print('\n── read() ───────────────────────────────────────')
with open(filepath, 'r', encoding='utf-8') as f:
    content = f.read()
print(content)

# READ LINE BY LINE
print('── readline() loop ──────────────────────────────')
with open(filepath, 'r') as f:
    while True:
        line = f.readline()
        if not line:
            break
        print(f'  {repr(line)}')

# READLINES
print('\n── readlines() ──────────────────────────────────')
with open(filepath, 'r') as f:
    all_lines = f.readlines()
print(f'  {len(all_lines)} lines read')
print(f'  First : {all_lines[0].rstrip()}')
print(f'  Last  : {all_lines[-1].rstrip()}')

# DIRECT ITERATION (most Pythonic for large files)
print('\n── Direct iteration ─────────────────────────────')
with open(filepath, 'r') as f:
    for i, line in enumerate(f, 1):
        print(f'  {i}: {line.rstrip()}')

os.remove(filepath)
````

### Program 19 – Append, r+ Mode & File Positions (tell/seek)

````python
import os

# Program 19: File Modes & Positions
# ─────────────────────────────────────────────────────────────

filepath = '/tmp/positions.txt'

with open(filepath, 'w') as f:
    f.write('ABCDEFGHIJ')

# Append mode
with open(filepath, 'a') as f:
    f.write('KLMNOPQRST')

with open(filepath) as f:
    print(f'After append: "{f.read()}"')

# tell() and seek()
print('\n── tell() & seek() ──────────────────────────────')
with open(filepath, 'r') as f:
    print(f'  Initial pos   : {f.tell()}')
    data = f.read(5)
    print(f'  read(5) = "{data}"  pos: {f.tell()}')
    f.seek(0)
    print(f'  seek(0)       pos: {f.tell()}')
    f.seek(0, 2)              # seek to end
    print(f'  seek(0,2)=end pos: {f.tell()} (= file size)')
    f.seek(5)
    print(f'  seek(5) read  : "{f.read()}"')

# r+ (read & write without truncating)
print('\n── r+ mode ──────────────────────────────────────')
with open(filepath, 'r+') as f:
    original = f.read()
    print(f'  Before: "{original}"')
    f.seek(0)
    f.write('***')    # overwrite first 3 bytes
with open(filepath) as f:
    print(f'  After : "{f.read()}"')

# w+ mode
print('\n── w+ mode (create + read/write) ────────────────')
with open(filepath, 'w+') as f:
    f.write('Hello w+ mode')
    f.seek(0)
    print(f'  Read back: "{f.read()}"')

# a+ mode
print('\n── a+ mode (append + read) ──────────────────────')
with open(filepath, 'a+') as f:
    f.write(' [appended]')
    f.seek(0)
    print(f'  Full file: "{f.read()}"')

os.remove(filepath)
print('\n(temp file cleaned up)')
````

### Program 20 – Binary File I/O with struct

````python
import struct, os

# Program 20: Binary File I/O
# ─────────────────────────────────────────────────────────────

FMT = 'I10sf'           # unsigned-int, 10-byte string, float
REC_SIZE = struct.calcsize(FMT)
bin_path = '/tmp/records.bin'
print(f'Record format: {FMT}   size: {REC_SIZE} bytes')

# Write records
records = [(1,'Alice',95.5),(2,'Bob',87.0),(3,'Carol',92.3),(4,'Dave',78.9)]
print('\n── Writing records ──────────────────────────────')
with open(bin_path, 'wb') as f:
    for rec_id, name, score in records:
        name_bytes = name.encode('utf-8').ljust(10, b'\x00')[:10]
        packed = struct.pack(FMT, rec_id, name_bytes, score)
        f.write(packed)
        print(f'  Wrote: ID={rec_id} Name={name:<6} Score={score}')

file_size = os.path.getsize(bin_path)
print(f'\nTotal: {file_size} bytes ({len(records)} records x {REC_SIZE} B)')

# Read records sequentially
print('\n── Reading records ──────────────────────────────')
print(f'  {"ID":<4} {"Name":<10} {"Score":>6}')
print('  ' + '-' * 22)
with open(bin_path, 'rb') as f:
    while True:
        chunk = f.read(REC_SIZE)
        if not chunk:
            break
        rec_id, raw_name, score = struct.unpack(FMT, chunk)
        name = raw_name.rstrip(b'\x00').decode('utf-8')
        print(f'  {rec_id:<4} {name:<10} {score:>6.1f}')

# Random access
print('\n── Random access by record index ────────────────')
with open(bin_path, 'rb') as f:
    for idx in [2, 0, 3]:
        f.seek(idx * REC_SIZE)
        chunk = f.read(REC_SIZE)
        rec_id, raw_name, score = struct.unpack(FMT, chunk)
        name = raw_name.rstrip(b'\x00').decode('utf-8')
        print(f'  Record[{idx}]: id={rec_id} name={name} score={score:.1f}')

os.remove(bin_path)
print('\n(binary file cleaned up)')
````

### Program 23 – Complete File Manager Utility

````python
import os, shutil, pathlib, datetime
from collections import Counter

# Program 23: Mini File Manager
# ─────────────────────────────────────────────────────────────

def file_info(path):
    p = pathlib.Path(path)
    st = p.stat()
    return {
        'name'    : p.name,
        'kind'    : 'DIR' if p.is_dir() else (p.suffix.upper() or '.?'),
        'size'    : st.st_size,
        'modified': datetime.datetime.fromtimestamp(st.st_mtime).strftime('%Y-%m-%d %H:%M'),
    }

def list_directory(path, pattern='*'):
    p = pathlib.Path(path)
    entries = sorted(p.glob(pattern), key=lambda x: (x.is_file(), x.name.lower()))
    print(f'\nDirectory: {p.resolve()}')
    print(f'{"Name":<28} {"Kind":<8} {"Size":>8}  {"Modified"}')
    print('-' * 60)
    for entry in entries:
        info = file_info(entry)
        tag  = '[D]' if entry.is_dir() else '[F]'
        print(f'{tag} {info["name"]:<26} {info["kind"]:<8} {info["size"]:>8,}  {info["modified"]}')
    print(f'\nTotal: {len(entries)} items')

def count_lines(path):
    try:
        with open(path, 'r', errors='ignore') as f:
            return sum(1 for _ in f)
    except:
        return -1

def search_files(root, keyword, extension=None):
    found = []
    for p in pathlib.Path(root).rglob('*'):
        name_match = keyword.lower() in p.name.lower()
        ext_match  = (extension is None) or (p.suffix.lower() == extension.lower())
        if name_match and ext_match:
            found.append(p)
    return sorted(found)

def disk_usage(root):
    total, count = 0, 0
    for p in pathlib.Path(root).rglob('*'):
        if p.is_file():
            total += p.stat().st_size
            count += 1
    return count, total

# ── Demo ──────────────────────────────────────────────────────
DEMO = '/tmp/fm_demo'
os.makedirs(f'{DEMO}/docs', exist_ok=True)
os.makedirs(f'{DEMO}/src',  exist_ok=True)

content = {
    f'{DEMO}/readme.txt'     : 'Project README\n' * 3,
    f'{DEMO}/config.ini'     : '[settings]\nkey=value\n',
    f'{DEMO}/docs/notes.txt' : 'Meeting notes\n' * 10,
    f'{DEMO}/docs/report.md' : '# Report\n\nContent.\n' * 5,
    f'{DEMO}/src/main.py'    : 'def main():\n    pass\n\nmain()\n',
    f'{DEMO}/src/utils.py'   : '# Utilities\n\ndef helper():\n    return 42\n',
}
for path, data in content.items():
    with open(path, 'w') as f: f.write(data)

list_directory(DEMO)

print('\n── Search for .txt files ─────────────────────────')
results = search_files(DEMO, '', '.txt')
for r in results:
    lines = count_lines(r)
    print(f'  {str(r.relative_to(DEMO)):<35} ({lines} lines)')

print('\n── Disk Usage ───────────────────────────────────')
file_count, total_bytes = disk_usage(DEMO)
print(f'  Files : {file_count}')
print(f'  Total : {total_bytes:,} bytes  ({total_bytes/1024:.2f} KB)')

print('\n── Extension Breakdown ──────────────────────────')
exts = Counter(p.suffix.lower() for p in pathlib.Path(DEMO).rglob('*') if p.is_file())
for ext, cnt in sorted(exts.items()):
    print(f'  {ext or "no-ext":<12}: {cnt} file(s)')

shutil.rmtree(DEMO)
print('\n(Demo cleaned up)')
````
