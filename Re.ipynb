Dot (.) – Any Single Character

```python
text = "cat cot cut act"

result = re.findall(r"c.t", text)

print(result)
```

`r"c.t"` means:

* c → first character must be c
* . → any single character
* t → last character must be t

So the pattern matches three-character words that start with c and end with t.

```python
text = "cat cot cut act"

result = re.findall(r"..t", text)

print(result)
```

`r'\b\d{2}\b'`

the \b represents a word boundary.

A word boundary is a position where:

* A word character (a-z, A-Z, 0-9, _) meets a non-word character (space, punctuation, start/end of string, etc.).

Breakdown

`\b      # Start word boundary`

`\d{2}   # Exactly 2 digits`

`\b      # End word boundary`

This pattern matches only standalone 2-digit numbers.

```python
import re

text = "I am 25 years old. Room number is 123. Code 45."

result = re.findall(r'\b\d{2}\b', text)
print(result)
```

```python
#without word boundary
result = re.findall(r'\d{2}', text)
print(text)
print(result)
```
